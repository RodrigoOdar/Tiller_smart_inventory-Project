```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Filter data for id_store 7304 and 8291
df_7304 = df_final[df_final['id_store_x'] == 7304]
df_8291 = df_final[df_final['id_store_x'] == 8291]

# Convert 'date_opened' to datetime
df_7304['date_opened'] = pd.to_datetime(df_7304['date_opened'])
df_8291['date_opened'] = pd.to_datetime(df_8291['date_opened'])

# Resample to monthly frequency and sum the billing
df_7304_monthly = df_7304.resample('M', on='date_opened')['m_total_price_inc_vat'].sum().reset_index()
df_8291_monthly = df_8291.resample('M', on='date_opened')['m_total_price_inc_vat'].sum().reset_index()

# Calculate cumulative billing
df_7304_monthly['cumulative_billing'] = df_7304_monthly['m_total_price_inc_vat'].cumsum()
df_8291_monthly['cumulative_billing'] = df_8291_monthly['m_total_price_inc_vat'].cumsum()

# Calculate total cumulative billing and total operational days for each store
total_cumulative_billing_7304 = df_7304_monthly['cumulative_billing'].iloc[-1]
total_cumulative_billing_8291 = df_8291_monthly['cumulative_billing'].iloc[-1]
total_operational_days_7304 = (df_7304_monthly['date_opened'].iloc[-1] - df_7304_monthly['date_opened'].iloc[0]).days
total_operational_days_8291 = (df_8291_monthly['date_opened'].iloc[-1] - df_8291_monthly['date_opened'].iloc[0]).days

# Calculate the ratios without decimals
ratio_7304 = int(total_cumulative_billing_7304 / total_operational_days_7304)
ratio_8291 = int(total_cumulative_billing_8291 / total_operational_days_8291)

# Create the plot
fig, ax1 = plt.subplots(figsize=(12, 6))

# Primary y-axis: Monthly Billing
sns.lineplot(x='date_opened', y='m_total_price_inc_vat', data=df_7304_monthly, label='Store 7304', ax=ax1)
sns.lineplot(x='date_opened', y='m_total_price_inc_vat', data=df_8291_monthly, label='Store 8291', ax=ax1)
ax1.set_ylabel('Monthly Billing')
ax1.legend(loc='upper left')

# Secondary y-axis: Cumulative Billing
ax2 = ax1.twinx()
sns.lineplot(x='date_opened', y='cumulative_billing', data=df_7304_monthly, label='Store 7304 Cumulative', ax=ax2, linestyle='--')
sns.lineplot(x='date_opened', y='cumulative_billing', data=df_8291_monthly, label='Store 8291 Cumulative', ax=ax2, linestyle='--')
ax2.set_ylabel('Cumulative Billing')
ax2.legend(loc='upper right')

# Add the labels for the ratios
plt.text(0.05, 0.5, f'Store 7304 Ratio: {ratio_7304}', transform=ax1.transAxes, fontsize=12,
         verticalalignment='center', bbox=dict(facecolor='white', alpha=0.8))
plt.text(0.05, 0.45, f'Store 8291 Ratio: {ratio_8291}', transform=ax1.transAxes, fontsize=12,
         verticalalignment='center', bbox=dict(facecolor='white', alpha=0.8))
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer


# Seleccionar los predictores y la variable objetivo
predictors = ['m_amount',	'm_nb_customer',	'm_quantity',	'day_of_week',	'month']

target = 'm_total_price_inc_vat'

# Separar los datos de entrada y salida para la tienda 7304
X_7304 = df_final_7304[predictors]
y_7304 = df_final_7304[target]

# Separar los datos de entrada y salida para la tienda 8291
X_8291 = df_final_8291[predictors]
y_8291 = df_final_8291[target]

# Dividir los datos de la tienda 7304 en entrenamiento y prueba
X_train, X_val, y_train, y_val = train_test_split(X_7304, y_7304, test_size=0.2, random_state=42)

# Identificar las columnas numéricas y categóricas
numeric_features = ['m_amount',	'm_nb_customer',	'm_quantity',	'day_of_week',	'month']

# Preprocesamiento: escalado para variables numéricas y codificación one-hot para categóricas
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features)])

# Crear el pipeline con el preprocesador y el modelo de RandomForest
model_gpt = Pipeline(steps=[('preprocessor', preprocessor),
                        ('regressor', RandomForestRegressor(n_estimators=100, random_state=42))])

# Entrenar el modelo con los datos de la tienda 7304
model_gpt.fit(X_train, y_train)

# Evaluar el modelo en los datos de validación de la tienda 7304
y_val_pred = model_gpt.predict(X_val)
print("R^2 en datos de validación de la tienda 7304:", r2_score(y_val, y_val_pred))
print("Error cuadrático medio en datos de validación de la tienda 7304:", mean_squared_error(y_val, y_val_pred))

# Aplicar el modelo a los datos de la tienda 8291 para predecir el target
y_8291_pred = model_gpt.predict(X_8291)

# Evaluar el rendimiento en los datos de la tienda 8291
print("R^2 en tienda 8291:", r2_score(y_8291, y_8291_pred))
print("Error cuadrático medio en tienda 8291:", mean_squared_error(y_8291, y_8291_pred))

# Crear un DataFrame con las predicciones y valores reales para la tienda 8291
results_RF8291 = pd.DataFrame({'Predicted': y_8291_pred, 'Actual': y_8291})
print(results_RF8291.head())
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from scipy.stats import randint

# Seleccionar los predictores y la variable objetivo
predictors = ['m_amount',	'm_nb_customer',	'm_quantity',	'day_of_week',	'month']

target = 'm_total_price_inc_vat'

# Separar los datos de entrada y salida para la tienda 7304
X_7304 = df_final_7304[predictors]
y_7304 = df_final_7304[target]

# Separar los datos de entrada y salida para la tienda 8291
X_8291 = df_final_8291[predictors]
y_8291 = df_final_8291[target]

# Dividir los datos de la tienda 7304 en entrenamiento y prueba
X_train, X_val, y_train, y_val = train_test_split(X_7304, y_7304, test_size=0.2, random_state=42)

# Identificar las columnas numéricas y categóricas
numeric_features = ['m_amount',	'm_nb_customer',	'm_quantity',	'day_of_week',	'month']

# Preprocesamiento: escalado para variables numéricas
preprocessor = ColumnTransformer(
    transformers=[('num', StandardScaler(), numeric_features)]
)

# Definir el modelo de Gradient Boosting
model = GradientBoostingRegressor(random_state=42)

# Crear el pipeline con el preprocesador y el modelo
pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                           ('regressor', model)])

# Definir la búsqueda de hiperparámetros
param_distributions = {
    'regressor__n_estimators': randint(50, 200),
    'regressor__learning_rate': [0.01, 0.05, 0.1, 0.2],
    'regressor__max_depth': randint(3, 10),
    'regressor__min_samples_split': randint(2, 10),
    'regressor__min_samples_leaf': randint(1, 10)
}

# Configurar la búsqueda aleatoria de hiperparámetros con validación cruzada
random_search = RandomizedSearchCV(
    pipeline,
    param_distributions=param_distributions,
    n_iter=20,
    scoring='r2',
    cv=3,
    random_state=42,
    n_jobs=-1
)

# Entrenar el modelo con los datos de la tienda 7304 y búsqueda de hiperparámetros
random_search.fit(X_train, y_train)

# Obtener el mejor modelo
best_model = random_search.best_estimator_

# Evaluar el mejor modelo en los datos de validación de la tienda 7304
y_val_pred = best_model.predict(X_val)
print("Mejor R^2 en datos de validación de la tienda 7304:", r2_score(y_val, y_val_pred))
print("Mejor Error cuadrático medio en datos de validación de la tienda 7304:", mean_squared_error(y_val, y_val_pred))

# Aplicar el modelo a los datos de la tienda 8291 para predecir el target
y_8291_pred = best_model.predict(X_8291)

# Evaluar el rendimiento en los datos de la tienda 8291
print("Mejor R^2 en tienda 8291:", r2_score(y_8291, y_8291_pred))
print("Mejor Error cuadrático medio en tienda 8291:", mean_squared_error(y_8291, y_8291_pred))

# Crear un DataFrame con las predicciones y valores reales para la tienda 8291
results_GB8291 = pd.DataFrame({'Predicted': y_8291_pred, 'Actual': y_8291})
print(results_GB8291.head())
```
